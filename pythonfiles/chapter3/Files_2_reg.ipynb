```python
#Read number of files in the folder
import os

folder_path = input("Enter the folder path: ")

# Dictionary to store file types (e.g., '.txt': 5)
file_types = {}
total_files = 0

if os.path.exists(folder_path) and os.path.isdir(folder_path):
    # List all items in the directory
    for item in os.listdir(folder_path):
        item_path = os.path.join(folder_path, item) #join the file path
        
        # Ensure we are only counting files, not subfolders
        if os.path.isfile(item_path):
            total_files += 1
            
            # Split the file name to get the extension
            extension = os.path.splitext(item)[1].lower()
            
            # If no extension, label it 'No Extension'
            if extension == "":
                extension = "No Extension"
                
            # Increment the count for this specific type
            file_types[extension] = file_types.get(extension, 0) + 1

    # Output results
    print("-" * 30)
    print(f"Total Files Found: {total_files}")
    print("-" * 30)
    print("Breakdown by Type:")
    for ext, count in file_types.items():
        print(f"{ext}: {count}")

else:
    print("Invalid folder path. Please check the path and try again.")
```
