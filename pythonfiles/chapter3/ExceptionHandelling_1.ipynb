```python
# unknown Error 
l1=[10,12,13]
try:
    v1=l1[10]
except(Exception ): #when we do not know what type of error
    print("unknown error")
else:
    print(v1)
```

```python
# unknown Error  better way to print actual error
l1=[10,12,13]
try:
    v1=l1[10]
except Exception as er: #when we do not know what type of error
    print(er.args)
    print("unknown error")
else:
    print(v1)

```

```python
# unknown Error  better way to print actual error. here you get type of the error, in below its "Index Error"
l1=[10,12,13]
try:
    v1=l1[10]
except Exception as er: #when we do not know what type of error
    print(er.with_traceback)
    print("unknown error")
else:
    print(v1)

```
