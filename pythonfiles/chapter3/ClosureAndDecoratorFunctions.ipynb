```python
def my_decorator(func):
    def wrapper():
        print("Something is happening before the function is called.")
        func()
        print("Something is happening after the function is called.")
    return wrapper

def say_whee():
    print("Whee!")

say_whee = my_decorator(say_whee)()
say_whee
```

```python
def decorator(func):
    def wrapper():
        print("Before calling the function.")
        func()
        print("After calling the function.")
    return wrapper

@decorator # Applying the decorator to a function. @decorator syntax is a shorthand for greet = decorator(greet).
def greet():
    print("Hello, World!")
greet()
```

```python
def decorator_name(func):
    def wrapper(*args, **kwargs):
        print("Before execution")
        #print(func(*args, **kwargs))
        res=func(*args, **kwargs)
        print(res)
        print("After execution")
       # return result
    return wrapper

@decorator_name
def add(*numbers):
    sum=0
    for a in numbers:
        sum+=a
    return sum

add(5, 3, 10)
#print(result)
```

```python
is_even_odd = lambda x: "Even" if x % 2 == 0 else "Odd"
print(is_even_odd(7)) # Output: Odd
```

```python
# map function
def square(n):
    return n * n

numbers = [1, 2, 3, 4, 5]
squared_map = map(square, numbers)
# Convert the map object to a list
result_list = list(squared_map)
print(result_list)
```
