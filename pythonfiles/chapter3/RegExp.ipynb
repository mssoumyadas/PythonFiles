```python
#Reg Ex
import re
type(re) # module is .py file
```

```python
#re.match(pattern,string,[flag]) match method matches the string from the start of the string.
str1="python is better language than java for AI"
str2="I like python"
print(re.match("python",str1)) #matching 
print(re.match("python",str2))
print(re.match("Python",str1,re.I)) # ignore case flag used 
```

```python
#re.search(pattern,string,[flags]) search methods looks the entire string for the text.
str1="python is better language than java for AI"
print(re.search("python",str1)) #search 
print(re.search("java",str1))
print(re.search("JAVA",str1))
print(re.search("JAVA",str1,re.I))
```

```python
ans=input("Are you coming for class today ?")
if re.search("yes",ans,re.I): # only when it returns value it becomes string, oterwise its none. none is false in python
    print("welcome")
else:
    print("take rest")  
```

```python
#Pattern design with meta characters 
# Position Anchors ^ (first) ,$(last)
#Quantifiers(how many times) * -0 or more, + 1 or more,? 0or 1, {n,m} - n and m times
#[] (Square Brackets): Matches any one character inside the brackets
#(Pipe): The "OR" operator. (a|b)
#. (Dot): Matches any single character except a newline.
ans=input("Enter a number")
if re.search('[0123456789]',ans): # fdsfds5uoi would also match as it contains one number. [0-9] would work too. [9-0] - wont work
    print("matched")
else:
    print("no match") 
```

```python
ans=input("Enter a number")
if re.search('[0-9][0-9]',ans): # two digits
    print("matched")
else:
    print("no match") 
```

```python
ans=input("Enter a number")
if re.search('[0-9]{3,}',ans): 
    print("matched")
else:
    print("no match") 
```

```python
# ^ first 
ans=input("Enter a string")
if re.search('^[0-9][a-z]',ans): # checks first is digit or not
    print("matched")
else:
    print("no match") 
```

```python
# ^ and$  it will accept only numbers
ans=input("Enter a string")
if re.search('^[0-9]*$',ans): #* ensure any number of digits.
    print("matched")
else:
    print("no match") 
```

```python
# ^ within [] it will not accept  numbers
ans=input("Enter a string")
if re.search('[^0-9]*',ans): #* ensure any number of digits.
    print("matched")
else:
    print("no match") 
```

```python
# the Pipe
ans=input("Enter a laguage")
if re.search('^(odia|english|hindi|Tamil|telegu|kannada)$',ans,re.I): 
    print("matched")
else:
    print("no match") 
```

```python
# ^ and$  it will accept only numbers
ans=input("Enter a string")
if re.search('^[0-9]*$',ans): #* ensure any number of digits.
    print("matched")
else:
    print("no match") 
```
