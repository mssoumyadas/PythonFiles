```python
#sub	Replaces one or many matches with a string
import re
str1= 'Cats are independent animals. cats may not be as wise as dogs. Comparing cats and dogs wrt maintenance cats are waty better.'
str2= re.sub('cat','cute',str1,count=0,flags=re.I )
print(str1)
print(str2)       
```

```python
#split	Returns a list where the string has been split at each match
str1= 'Cats are independent animals.'
str2= str1.split() # split is a method from string
print(str1)
print(str2)
text = "apple;banana orange,grape"
# Splits on any of: semicolon, space, or comma
print(re.split(r'[; ,]', text)) # Output: ['apple', 'banana', 'orange', 'grape']
print(re.split(r'[; ,]',str1)) # Output: ['apple', 'banana', 'orange', 'grape']
```

```python
#findall	Returns a list containing all matches
str1= 'Cats are independent animals. cats may not be as wise as dogs. Comparing cats and dogs wrt maintenance cats are waty better.'
str2= re.findall('cat',str1,flags=re.I )
print(str1)
print(str2) 
print(len(str2))
```

```python
txt = "The rain in Spain"

#Check if "ain" is present at the beginning of a WORD:

x = re.findall(r"\bain", txt)

print(x)

if x:
  print("Yes, there is at least one match!")
else:
  print("No match")
```
