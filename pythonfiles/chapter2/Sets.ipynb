```python
#Sets
#no index  no slice
# read sets with For loop
s1={1,2,3,4,5}
s2={3,4,5,6,7}
print(s1-s2)
print(s2-s1)
print(s1&s2)
print(s1|s2)
print(s1^s2)
```

```python
L1=[10,20,30,40,50]
S1=set(L1) #set from list, it can be created from a string, range, list, tuple, dictionary and set
S1
```

```python
Dogs= {'a': 'Akita', 'b': 'beagle', 'c': 'Cocker Spaniel', 'd': 'daschund', 'e': 'English Bulldog', 'h': 'Huskey'}
Dog=set(Dogs)
print(Dog) # creates only with keys
```

```python
DogValue=set(Dogs.items()) 
print(DogValue) # taking both key value for set
```
