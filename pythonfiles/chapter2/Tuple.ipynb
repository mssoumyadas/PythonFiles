```python
#Tuple
T1= tuple("welcome") # from string
print(T1)
T2= tuple(range(5,50,5)) # from range
print(T2)
T3= tuple(['cat', 'dog', 'cockatiel', 'goat', 'Peace Lily']) # from List
print(T3)
T4=tuple((1,2,3,3)) #from tuple
print(T4)
```

```python
#Creating single element Tuple
T5=(100,)
#T6=tuple(100) can not create tuple in this way has to end with comma
print(type(T5))
T6= 12,13,14  # by default comma separated item without parenthesis are considered as Tuple
print(type(T6))
```

```python
#Tuple Methods
print(T4.count(3))
print(T2.count(5))
for i in T3: #index
    print(i)
print(T3.index("dog"))
T10=(1,2,3,4,5,6,7,4,3,5,3,4)
print(T10.index(3))
print(T10.index(3,3))
print(T10.index(3,9))
```

```python
#combination of List and Tuple
TL=([1,10],[2,20],[3,30])
print(TL[1])
print(TL[2][1])
```
