```python
#ListMethods
L1= ["cat", "dog", "cockatiel", "goat", "rooster"]
# dir(L1) to see all the available methods of list
print(*dir(L1), sep=', ') # iterable unpacking using"*"
```

```python
L1.append("Siamese fighting fish")
print(L1)

```

```python
L2 =L1.copy()
L1.clear() #Clear doesnt take any argument
print(L1) 
print(L2)
```

```python
print(L2.count('cat'))
```

```python
T1=("Money Plant", "Peace Lily")
L2.extend(T1) #extend can combine two lists, List and touple. Basically iterable objects
print(L2)
```

```python
print(L2.index('dog'))
#print(L2.insert( 6,"Snake plant")) Question why this displays none? The reason is insert doesnt return anything hence none
(L2.insert( 6,"Snake plant"))
print(L2)
```

```python
# remove methods "del", "Pop" and "remove"
L1=[1,2,3,4,5]
del L1[3] # postition 4 deleted 
print(L1)
# delete few elements
L2 = ['cat', 'dog', 'cockatiel', 'goat', 'rooster', 'Money Plant', 'Peace Lily']
del L2[4:6]
print(L2)

#Remove Method
L1.remove(5)
print(L1)
L1=[1,2,3,4,5]
L1.pop()
print(L1)
L1.pop(1)
print(L1)
```
