```python
#markersize, markerfacecolor, markeredgecolour, linestyle
import matplotlib.pyplot as plt
import numpy as np  
x=['jan','feb','mar','apr','may','June']
sunnydays=[20,20,30,25,30,15]
winddays=[28,28,25,20,15,20]
linestyle =dict (marker='.',markersize='20',mfc='red',mec='green',linestyle='solid',linewidth='2',alpha=0.8)
fontstyle= dict (family='Arial',fontsize='20',fontweight='bold')

#plt.plot(x,sunnydays,marker='*',markersize='20',mfc='red',mec='green',linestyle='dashed',linewidth='4',color='blue')
plt.plot(x,sunnydays,color='green',label='sunny days',**linestyle)
plt.plot(x,winddays,color='blue',label='winddays',**linestyle)
#plt.title('Green Energy',family='Arial',fontsize='20',fontweight='bold',color='#42f545')
plt.title('Green Energy',color='#42f545',**fontstyle)
plt.xlabel('Months',color='#4248f5',**fontstyle)
plt.ylabel('Days of Sun/Wind',color='#4248f5',**fontstyle)
plt.subplots_adjust(bottom=0.2, left=0.15)
#plt.legend(loc="best")
plt.legend(loc="lower left")
plt.grid()
plt.show()
#plt.xticks(x)
```
