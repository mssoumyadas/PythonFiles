```python
import matplotlib.pyplot as plt
import numpy as np  
diet=np.array(["Fruits","vegitable","grains","legumes"])
quantity= np.array([20,40,30,20])
plt.bar(diet,quantity)
# plt.barh(diet,quantity)  Horizontal chart
plt.xlabel("Ingredients")
plt.ylabel("Quantity")
plt.title("Diet Plan")

```

```python
diet=np.array(["Fruits","vegitable","grains","legumes"])
quantity= np.array([20,40,30,20])
color=['red','green','brown','yellow']
plt.pie(quantity,labels=diet,autopct="%1.1f%%",colors=color,explode=[0,0.1,0,0],shadow=True,startangle=230)
plt.title("My Daily Diet")
plt.show()
```
