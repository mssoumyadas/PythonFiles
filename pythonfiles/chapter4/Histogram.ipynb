```python
#histogram for school score distributions
import matplotlib.pyplot as plt
import numpy as np  
scores= np.random.normal(loc=80,scale=10,size=100)
scores=np.clip(scores,0,100)
plt.hist(scores,bins=8,color='lightgreen',edgecolor='black')
plt.title("Exam Score")
plt.xlabel("Scores")
plt.ylabel("Students")
#plt.hist(scores)
```
