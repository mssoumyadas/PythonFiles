```python
import matplotlib.pyplot as plt
import numpy as np  
# Figure: The entire canvas
#Axis : A single plot
# print(plt.subplots(2,2))
x=np.array([1,2,3,4,5])
figure, axis= plt.subplots(2,2)
axis[0,0].plot(x,x*2)
axis[0,0].set_title("x*2")
axis[0,1].plot(x,x**2)
axis[0,1].set_title("x**2")
axis[0,0].plot(x,x*2)
axis[0,0].set_title("x*2")
axis[1,0].plot(x,x*5)
axis[1,0].set_title("x*5")
axis[1,1].plot(x,x**3)
axis[1,1].set_title("x**3")
plt.tight_layout() # Prevents overlapping of titles and labels
plt.show()
```
