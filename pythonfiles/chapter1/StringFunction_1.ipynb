```python
#StringFunction
sampleString="Hope the wars end. We live in harmony."
SampleGerman= "Grüß Gott! Good day! "
print(SampleGerman.casefold()) # works for non-engilish language
print(sampleString.count('e'))
print(sampleString.endswith('harmony.'))
print(sampleString.endswith('in harmony.'))
print(sampleString.expandtabs(50)) #  check expand tabs
print(sampleString.find('wars'));
print("My name is {0}, I'm {1}".format("John",36))
SampleDict1={"name": "Soumya" , "age": "40"}
SampleDict2={"name": "Arabinda" , "age": "43"}
txt = "Happy birthday {name} you are at {age}s!"
print(txt.format_map(SampleDict1))
print(txt.format_map(SampleDict2))
print(sampleString.index('wars')) #same as find. 
alnum="123Hello"
print(sampleString.isalnum())
print(alnum.isalnum())
alapha= "HelloWorld"
print(sampleString.isalpha())
print(alapha.isalpha())
```

```python
print(sampleString.isascii())
print(sampleString.isdecimal())
print(sampleString.isdigit())
print(sampleString.isidentifier()) 
print(sampleString.isprintable())
print(sampleString.isspace())
testTitle="Welcome Back"
print(sampleString.istitle())
print(testTitle.istitle())
TupleTest= ("hello","dearWorld","again")
print(sampleString.join(TupleTest)) #join can join list and Tuple
print("-".join(TupleTest)) # the string calling the method can act a separator
```

```python
#maketrans
# Example: Using a dictionary
mapping = {ord('a'): 'x', ord('b'): 'y', ord('c'): None} # None means delete
text = "abcdef"
translation_table = str.maketrans(mapping)
print(text.translate(translation_table))
# Example: Using two strings
str1 = "heiou"
str2 = "12345"
text = "hello world"
translation_table = str.maketrans(str1, str2)
print(text.translate(translation_table))

```
