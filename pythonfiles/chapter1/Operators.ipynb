```python
# bitwise 
a = 60
b = 13

print(f"a = {a} ({a:08b})")
print(f"b = {b} ({b:08b})\n")
c = a & b
print(f"a & b = {c} ({c:08b})")
```

```python
# comparison operators
x = 5
y = 3

print(x == y)
print(x != y)
print(x > y)
print(x < y)
print(x >= y)
print(x <= y)
```

```python
#Logical Operators
x=5
print(not(x > 3 and x < 10))
print(x > 0 and x < 10)
print(x < 5 or x > 10)
print (3 or 5)
print (3 and 5)
```

```python
#Identity Operators
x= 5
y="5"
print(x is y)
print(x is not y)
```
